```python
import numpy as np
import pandas as pd

df = pd.read_csv("dirty_cafe_sales.csv")
df_clean = df.copy()

invalid_placeholders = ["UNKNOWN", "ERROR", "unknown", "error", "", " "]
df_clean.replace(invalid_placeholders, np.nan, inplace=True)

item_price_map = {
    "Coffee": 2.0,
    "Tea": 1.5,
    "Sandwich": 4.0,
    "Salad": 5.0,
    "Cake": 3.0,
    "Cookie": 1.0,
    "Smoothie": 4.0,
    "Juice": 3.0,
}

price_item_map = {v: k for k, v in item_price_map.items()}

for col in ["Quantity", "Price Per Unit", "Total Spent"]:
    df_clean[col] = pd.to_numeric(df_clean[col], errors="coerce")

df_clean["Price Per Unit"] = df_clean["Price Per Unit"].fillna(
    df_clean["Item"].map(item_price_map)
)

df_clean["Item"] = df_clean["Item"].fillna(
    df_clean["Price Per Unit"].map(price_item_map)
)

df_clean["Quantity"] = df_clean["Quantity"].fillna(
    df_clean["Total Spent"] / df_clean["Price Per Unit"]
)

df_clean["Total Spent"] = df_clean["Quantity"] * df_clean["Price Per Unit"]

df_clean["Payment Method"] = df_clean["Payment Method"].fillna("Desconocido")
df_clean["Location"] = df_clean["Location"].fillna("Desconocido")

df_clean["Transaction Date"] = pd.to_datetime(
    df_clean["Transaction Date"], errors="coerce"
)

df_clean.dropna(subset=["Transaction Date"], inplace=True)

df_clean.to_csv("clean_cafe_sales.csv", index=False)
print("¡Limpieza completada y archivo exportado exitosamente!")
```
